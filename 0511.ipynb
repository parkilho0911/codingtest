문제 : 여행가 A는 N x N크기의 정사각형 공간 위에 서 있다. 이 공간은 1 x 1 크기의 정사각형으로 나누어져 있다. 가장 왼쪽 위 좌표는 (1,1)이며, 가장 오른쪽 아래 좌표는 (N,N)에 해당한다. <br> 
여행가 A는 상, 하, 좌, 우 방향으로 이동할 수 있으며, 시작 좌표는 항상(1,1)이다. <br>
우리 앞에는 여행가 A가 이동할 계획이 적힌 계획서가 놓여있다. <br>
계획서에는 하나의 줄에 띄어쓰기를 기준으로 하여 L,R,U,D중 하나의 문자가 반복적으로 적혀 있다.<br>
각 문자의 의미는 다음과 같다.<br>
L:왼쪽으로 한 칸 이동 <br>
R : 오른쪾으로 한 칸 이동<br>
U : 위로 한 칸 이동<br>
D : 아래로 한 칸 이동<br>
이때 여행가 A가 N x N 크기의 정사각형 공간을 벗어나는 움직임은 무시된다. <br>
예를 들어 (1,1)의 위치에서 L 혹은 U를 만나면 무시된다. <br>
다음은 N=5인 지도와 계획서이다<br>
이 경우 6개의 명령에 따라서 여향가가 움직이게 되는 위치는 순서대로 (1,2),(1,3),(1,4),(1,4),(2,4),(3,4)이므로, 최종적으로 여행가A가 도착하게 되는 곳의 좌표는 (3,4)이다.<br>
다시 말해 3행 4열의 위치에 해당하므로 (3,4)라고 적는다. <br>
계획서가 주어졌을 때 여행가 A가 최종적으로 도착할 지점의 좌표를 출력하는 프로그램을 작성하시오.

입력 조건 : 첫쨰 줄에 공간의 크기를 나타내는 N이 주어진다 <br>
            둘쨰 줄에 여행가 A가 이동할 계획서 내용이 주어진다<br>
출력 조건 : 첫째 줄에 여행가 A가 최종적으로 도착할 지점의 좌표(X,Y)를 공백으로 구분하여 출력한다.

입력 예시 : <br>
5<br>
R R R U D D<br>
출력 예시 : <br>
3 4

```python
#내가 작성한 코드
N = int(input())
move = map(input().split())
x,y=1
r,l,u,d = 1
for i in move():
    if move == l:
        x -=1
    elif move == r:
        x +=1
    elif move == u:
        y +=1
    else:
        y-=1

print(x,y)
```

```python
#내 코드를 제미나이가 수정한 것
N = int(input())
moves = input().split() # 1. map 제거: 그냥 리스트로 받기
x, y = 1, 1 # 2. 변수 짝 맞춰주기

for i in moves: # 3. 괄호 제거: 리스트에서 하나씩 꺼내기 (i에는 'R', 'L' 등이 담김)
    
    # 💡 핵심: 진짜 x, y를 바로 바꾸지 않고, 미리 가볼 임시 좌표(nx, ny)를 만듭니다.
    nx, ny = x, y 
    
    # 4. i와 진짜 문자열 비교하기 (좌표 x, y 변경 기준 맞추기)
    if i == 'L':
        ny -= 1 # 왼쪽은 열(y)이 감소
    elif i == 'R':
        ny += 1 # 오른쪽은 열(y)이 증가
    elif i == 'U':
        nx -= 1 # 위쪽은 행(x)이 감소
    elif i == 'D':
        nx += 1 # 아래쪽은 행(x)이 증가

    # 5. 공간 방어: 임시로 가본 곳이 지도(1~N)를 벗어난다면?
    if nx < 1 or ny < 1 or nx > N or ny > N:
        continue # 아무것도 하지 않고 다음 명령(moves)으로 넘어감
        
    # 지도를 벗어나지 않았다면, 진짜 위치(x, y)를 갱신해 줌
    x, y = nx, ny

print(x, y)
```

```python
#나동빈 코드
n = int(input())
x,y = 1,1
plans = input().split()

#L,R,U,D
dx = [0,0,-1,1]
dy = [-1,1,0,0]
move_types = ['l','r','u','d']

for plan in plans:
    for i in range(len(move_types)):
        if plan == move_types[i]:
            nx = x + dx[i]
            ny = y + dy[i]
        
    if nx < 1 or ny < 1 or nx > n or ny > n:
        continue
    x,y = nx, ny

print(x,y)
```

# 게임 개발

현민이는 게임 캐릭터가 맵 안에서 움직이는 시스템을 개발 중이다. <br>
캐릭터가 있는 장소는 1 X 1 크기의 정사각형으로 이뤄진 N x M 크기의 직사각형으로, 각각의 칸은 육지 또는 바다이다. 캐릭 터는 동서남북 중 한 곳을 바라본다.<br>
맵의 각 칸은 (A. B)로 나타낼 수 있고, A는 북쪽으로부터 떨어진 칸의 개수, B는 서쪽으로부터 떨 어진 칸의 개수이다. 캐릭터는 상하좌우로 움직일 수 있고, 바다로 되어 있는 공간에는 갈 수 없다.<br>
캐릭터의 움직임을 설정하기 위해 정해 놓은 매뉴얼은 이러하다.<br>
1. 현재 위치에서 현재 방향을 기준으로 왼쪽 방향 (반시계 방향으로 90도 회전한 방향)부터 차례대로 갈 곳을 정한다.<br>
2. 캐릭터의 바로 왼쪽 방향에 아직 가보지 않은 칸이 존재한다면, 왼쪽 방향으로 회전한 다음 왼쪽으로 한 칸을 전진한다. <br>왼쪽 방향에 가보지 않은 칸이 없다면, 왼쪽 방향으로 회전만 수행하고 1단계로 돌아간다.<br>
3. 만약 네 방향 모두 이미 가본 칸이거나 바다로 되어 있는 칸인 경우에는, 바라보는 방향을 유지한 채로 한 칸 뒤로 가고 1단계로 돌아간다. <br>단, 이때 뒤쪽 방향이 바다인 칸이라 뒤로 갈 수 없는 경우에는 움직임을 멈춘다.<br>
현민이는 위 과정을 반복적으로 수행하면서 캐릭터의 움직임에 이상이 있는지 레스트하려고 한다.<br>
매뉴얼에 따라 캐릭터를 이동시킨 뒤에, 캐릭터가 방문한 칸의 수를 출력하는 프로그램을 만드시오.<br>

입력 조건 : <br>
• 첫째 줄에 맵의 세로 크기 N과 가로 크기 M을 공백으로 구분하여 입력한다. (3 S N. M s50)<br>
• 둘째 줄에 게임 캐릭터가 있는 칸의 좌표 (A. B)와 바라보는 방향 d가 각각 서로 공백으로 구분하여<br>
주어진다. 방향 d의 값으로는 다음과 같이 4가지가 존재한다.<br>
0 : 북쪽<br>
1 : 동쪽 <br>
2 : 남쪽 <br>
3 : 서쪽 <br>
셋째 줄부터 맵이 육지인지 바다인지에 대한 정보가 주어진다. <br>
N개의 줄에  맵의 상태가 북쪽부터 남쪽 순서대로, 각 줄의 데이터는 서쪽부터 동쪽 순서대로 주어진다. 맵의 외곽은 항상 바다로 되어 있다.<br>
0 : 육지<br>
1 : 바다<br>
처음에 게임 캐릭터가 위치한 칸의 상태는 항상 육지이다.<br>
출력 조건 : <br>
첫째 줄에 이동을 마친 후 캐릭터가 방문한 칸의 수를 출력한다<br>
입력 예시 : <br>
4 4 <br>
1 1 0<br>
1 1 1 1<br>
1 0 0 1<br>
1 1 0 1<br>
1 1 1 1<br>
출력 예시 : <br>
3

```python
n,m = map(int, input().split()) # 가로,세로 입력
x,y,d = map(int, input().split()) # 캐릭터 좌표, 방향

see = {0:"북쪽", 1:"동쪽", 2:"남쪽", 3:"서쪽"}
land = {0:"육지", 1:"바다"}

move = { 0:(-1,0), 1:(0,1), 2:(1,0), 3:(0,-1)}


d_map = [[0] * m for _ in range(n)]
d_map[x][y] = 1

array = []
for i in range(n):
    array.append(list(map(int, input().split())))

count = 1
turn_time = 0

while True:
    d -=1
    if d == -1:
        d = 3
    
    nx = x + move[d][0]
    ny = y + move[d][1]

    if d_map[nx][ny] == 0 and array[nx][ny] == 0:
        d_map[nx][ny] = 1
        x, y = nx, ny
        count += 1
        turn_time = 0
        continue
    else:
        turn_time += 1
    
    # 4면이 모두 막혔다면?
    if turn_time == 4:
        # 후진하기 (더했던 좌표를 빼주면 됩니다)
        nx = x - move[d][0]
        ny = y - move[d][1]
        
        if array[nx][ny] == 0: # 뒤가 육지면 후진
            x, y = nx, ny
            turn_time = 0
        else: # 뒤가 바다(1)면 게임 오버
            break

print(count)
```
