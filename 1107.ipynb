# 어제 다 못푼 그리디 문제 2개 풀기

```python
#1860번 진기의 최고급 붕어빵
#n명의 사람이 자격을 얻는다
#0초부터 m초의 시간으로 k개의 붕어빵을 만듬
t = int(input())

for test_case in range(1,t+1):
    n,m,k = map(int, input().split())
    #n은 사람 수, m은 만드는 시간, k는 붕어빵 갯수
    x = list(map(int, input().split()))
    #x는 도착하는 시간
    count = 0 #빵 개수
    i = 0 #시간초
    served = 0 #받아간 사람
    answer = "Possible"

    # 1번 붕어빵을 m초에 걸려서 k개 만든것을 어디에 저장해 놔야한다
    # 2번 손님의 도착 시간을 보고 k개를 나눠줄 수 있는지 확인한다
    # 3번 만약 손님에게 붕어를 주지 못하면 impossible을 주고 다 줄 수 있다면 possible을 출력한다
    while served < n:
        if i > 0 and i % m == 0:
            count +=k

        for j in x:
            if j==i:
                if count == 0:
                    answer = "Impossible"
                    break
                count -=1
                served +=1

        if answer == "Impossible":
            break
        i+=1
    print(f"#{test_case} {answer}")
```

```python
#7964번 부먹왕국의 차원 관문
t = int(input())

for test_case in range(1,t+1):
    n,d = map(int, input().split())
    a = list(map(int, input().split()))
    count = 0
    answer = 0

    for i in range(n):
        if a[i] == 0:
            count +=1
        else:
            count = 0

        if count == d:
            answer +=1
            count = 0

    print(f"#{test_case} {answer}")

```

```python
#1860번 복습 진기의 최고급 붕어빵
t = int(input())

for test_case in range(1,t+1):
    n,m,k = map(int, input().split())
    #n은 사람, m은 시간, k는 개수
    a = list(map(int, input().split()))
    #a는 사람이 도착하는 시간
    people = 0 #받은 사람
    count = 0 #붕어 갯수
    i = 0 #시간
    answer = "Possible"
    while people < n:
        if i %m==0 and i > 0:
            count += k

        for j in a:
            if j==i:
                if count==0:
                    answer = "Impossible"
                    break
                count -=1
                people +=1
        if answer == "Impossible":
            break
        i+=1
    print(f"#{test_case} {answer}")
```

```python
#7964번 부먹왕국의 차원 관문
t = int(input())

for test_case in range(1, t+1):
    n,d = map(int, input().split())
    a = list(map(int, input().split()))
    count = 0
    b = 0

    for i in a:
        if i == 0:
            b +=1

        if i == 1:
            b = 0

        if b == d:
            count +=1
            b = 0

    print(f"#{test_case} {count}")

```

```python
#1206번 [S/W 문제해결 기본] 1일차 - View
#빌딩의 좌,우의 공간이 2칸 이상이 되어야지만 count할 수 있다.
t = 10

for test_case in range(1,t+1):
    n = int(input())
    a = list(map(int, input().split())) # 건물의 높이
    #a[0]의 높이에서 b와 c의 높이가 얼마나 인지 확보가 되어야지 count할 수 있다.
    total = 0

    for i in range(2,len(a) -2):
        count = 0
        #10번을 반복해. 0번일때 오른쪽 2개 건물을 확인한다.
        #리스트가 2일때 확인해야하는거는 0,1그리고 3,4이렇게 총 4개를 비교해야한다.
        #그렇지만 0,1 n, n-1일때는 어떻게 해야할까??
        #i = 2일때를 만들어보자
        if a[i] > a[i-1] and a[i] > a[i-2]:
            count +=1

        if a[i] > a[i+1] and a[i] > a[i+2]:
            count +=1

        if count == 2:
            total += a[i] - max(a[i-1], a[i-2], a[i+1], a[i+2])


    print(f"#{test_case} {total}")
```

```python
#1209번 [S/W 문제해결 기본] 2일차 - Sum
t = 10

for test_case in range(1, t+1):
    n = int(input())
    x = []
    for i in range(100):
        x.append(list(map(int, input().split())))
    total = []
    total_sum = 0
    answer = 0
    for row in range(len(x)): #이게 행, 열을 구한 값
        total_row = 0
        total_col = 0
        for col in range(len(x)):
            total_col += x[col][row] #열의 합
            total_row += x[row][col] #행의 합
        total.append(total_col)
        total.append(total_row)

    for j in range(100):
        total_sum += x[j][j]
    total.append(total_sum)
    total_sum = 0

    for j in range(100):
        total_sum += x[j][99-j]
    total.append(total_sum)


    max_total = max(total)
    answer = max_total

    print(f"#{test_case} {answer}")
```
