```python
#1970번 복습 쉬운 거스름돈
t = int(input())
for test_case in range(1, t+1):
    n = int(input())
    money = [50000,10000,5000,1000,500,100,50,10]
    score = []
    count = 0
    for i in money:
        count = n//i
        score.append(count)
        n%=i

    print(f'#{test_case}')
    print(*score)
```

```python
#21425번 복습 +=
t = int(input())

for test_case in range(t):
    A,B,N = map(int, input().split())
    count = 0

    while max(A,B) <= N:
        if A < B:
            A +=B
        else:
            B +=A
        count +=1
    print(count)
```

```python
#1859번 백만 장자 프로젝트
t = int(input())  # 테스트케이스 개수

for test_case in range(1, t + 1):  # 각 테스트케이스 처리
    n = int(input())  # 날짜 수
    a = list(map(int, input().split()))  # 날짜별 가격

    total = 0  # 누적 이익
    max_a = a[-1]  # 마지막 날 가격을 판매 기준으로 설정

    for i in reversed(a):  # 마지막 날부터 첫날까지 가격을 확인
        if i > max_a:  # 더 높은 가격을 발견했다면
            max_a = i  # 앞선 날들에 적용할 판매 기준을 갱신

        elif i < max_a:  # 현재 가격보다 미래 판매 가격이 높다면
            total += max_a - i  # 한 개를 사고팔 때의 이익을 누적

    print(f"#{test_case} {total}")  # 최종 이익 출력
```

```python
# 4831번 전기버스
t = int(input())  # 테스트케이스 개수

for test_case in range(1, t + 1):
    # k: 최대 이동 거리, n: 종점, m: 충전소 개수
    k, n, m = map(int, input().split())

    # 충전소가 있는 정류장 번호
    stations = list(map(int, input().split()))

    current = 0  # 현재 출발 위치
    count = 0    # 충전 횟수

    # 현재 위치에서 종점까지 바로 갈 수 없다면 충전소를 찾음
    while current + k < n:
        next_stop = -1  # 아직 다음 충전소를 찾지 못한 상태

        # 갈 수 있는 가장 먼 위치부터 뒤쪽으로 확인
        for candidate in range(current + k, current, -1):
            if candidate in stations:  # 충전소가 있는 위치라면
                next_stop = candidate  # 다음 충전 위치로 선택
                break                  # 충전소 탐색 반복문 종료

        # 이동 가능한 범위에 충전소가 하나도 없다면
        if next_stop == -1:
            count = 0  # 도착할 수 없는 경우의 정답은 0
            break      # 버스 이동 반복문 종료

        # 찾은 충전소로 이동하여 충전
        current = next_stop
        count += 1

    # 테스트케이스별 결과 출력
    print(f"#{test_case} {count}")
```

```python
#1289번 원재의 메모리 복구하기
#힌트 후 해결
t = int(input())

for test_case in range(1,t+1):
    n = input().strip()
    total = []
    a = len(n)
    answer = [0]*a
    count = 0
    for i in n:
        total.append(int(i))

    for j in range(a):
        if answer[j] !=total[j]:
            for k in range(j,a):
                answer[k] = total[j]
            count +=1


    print(f"#{test_case} {count}")
```

```python
#4789번 성공적인 공연 기획
t = int(input())

for test_case in range(1,t+1):
    n = input().strip()
    total = []
    human = 0
    answer = 0

    for i in str(n):
        total.append(int(i))

    for j in range(len(total)):
        if total[j] >0:
            if human < j:
                need = j-human
                answer += need
                human += need

            human += total[j]

    print(f"#{test_case} {answer} ")

```

```python
#5201번 컨테이너 운반
t = int(input())

for test_case in range(1,t+1):
    n,m = map(int, input().split()) # 3 2
    #n은 컨테이너 수, m은 트럭의 수
    wi = list(map(int, input().split())) # 5 3 1
    #wi는 화물의 무게, ti는 적재 용량
    ti = list(map(int, input().split())) # 8 3
    wi.sort(reverse=True)
    ti.sort(reverse=True)
    total = 0

    for i in range(m):
        for j in range(len(wi)):
            if ti[i] >= wi[j]:
                total += wi[j]
                wi.pop(j)
                break

    print(f"#{test_case} {total}")

```

```python
#5202번 화물 도크
t = int(input())

for test_case in range(1,t+1):
    n = int(input())
    time = []
    for i in range(n):
        start, end = map(int, input().split())
        time.append([start, end])
    time.sort(key=lambda x:x[1])

    last_end = 0
    count = 0

    for i in range(n):
        start = time[i][0]
        end = time[i][1]

        if start >= last_end:
            count +=1
            last_end = end
    print(f"#{test_case} {count}")
```
