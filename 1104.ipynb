```python
#1948번 복습 날짜 계산기
t = int(input())

for start in range(1, t+1):
    m1,d1,m2,d2 = map(int, input().split())
    days = [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    count = 0

    if m1==m2:
        count += d2-d1+1
    else:
        if m1!=m2:
            #첫달 계산
            count +=days[m1]-d1+1

            #중간달 계산
            for i in range(m1+1, m2):
                count += days[i]
            
            #마지막달 계산
            count += d2
    
    print(f"#{start} {count}")
```

```python
#1204번 1일차 - 최빈수 구하기
t = int(input())
for start in range(1,t+1):
    n=int(input())
    count = [0]*101
    student = list(map(int, input().split()))

    for i in student:
        count[i]+=1

    max_count = max(count)
    answer = 0

    for i in range(101):
        if count[i]==max_count:
            answer = i #answer같은 정답을 넣을 변수를 하나 만들어주는게 좋음

    print(f"#{start} {answer}")    
```

```python
#1945번 복습 간단한 소인수 분해
t = int(input())

for start in range(1, t+1):
    n = int(input())
    insu = [0]*5

    while n>1:
        
        if n%11==0:
            insu[4]+=1
            n = n//11
        elif n%7==0:
            insu[3]+=1
            n = n//7
        elif n%5==0:
            insu[2]+=1
            n = n//5
        elif n%3==0:
            insu[1]+=1
            n = n//3
        elif n%2==0:
            insu[0]+=1
            n = n//2

    print(f"#{start}",*insu)
```

```python
#1940번 복습 가랏! RC카!
#틀렸음


t = int(input())

for start in range(1, t+1):
    n = int(input())
    speed = 0
    distance = 0

    for i in range(n):
        a,b = map(int, input().split())

        if a==1:
            speed +=b
        elif a==0:
            pass
        elif a==2:
            speed = max(0, speed-b)

        distance +=speed

    print(f"#{start} {distance}")
```

```python
#21425번 += 문제
t = int(input())

for i in range(t):
    a,b,n = map(int, input().split())
    
    count = 0
    while max(a,b) <= n:
        if a<b:
            a +=b
        else:
            b +=a
        count +=1

    print(count)
```
