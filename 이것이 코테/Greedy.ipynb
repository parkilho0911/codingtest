### 유튜브 동영상에서 나온 문제들

```python
#1이 될 때까지
n,k = map(int, input().split())
count = 0
while n>1:
    if n%k==0:
        n = n//k
        count +=1
    elif n==1:
        break
    elif n%k!=0:
        n-=1
        count +=1
print(count)
```

```python
#곱하기 혹은 더하기
n = input().strip()
a = []
total = 0
for i in str(n):
    a.append(int(i))

for j in a:
    if total ==0 or j < 2:
        total += j
    else:
        total *=j
print(total)
```

```python
n = int(input())
a = list(map(int, input().split()))
a.sort()
total = 0
count = 0

for i in a:
    total +=1
    if total >= i:
        count +=1
        total =0
print(total)
```

### 책에서 나오는 문제들

```python
#거스름돈
n = int(input())
coin = [500,100,50,10]
count = 0

for i in coin:
    count += n//i
    n%=i

print(count)
```

```python
#큰 수의 법칙
n,m,k = map(int, input().split())
a = list(map(int, input().split()))
a.sort(reverse = True)
total = 0
next_a = a[1]

for i in range(m):
    if (i+1)%(k+1)==0 and i!=0:
        total += next_a
    else :
        total += max(a)
print(total)
```

```python
#숫자 카드 게임
n,m = map(int, input().split())
a = []
score = []
for i in range(n):
    a.append(list(map(int, input().split())))

for i in range(n):
    score.append(min(a[i]))
print(max(score))
```
