```python
#상하좌우 문제
n = int(input())
move = [-1,1,-1,1] #L, R, U, D
a = list(map(str, input().split()))
answer = [1,1]

for i in range(len(a)):
    if a[i]== "L":
        if answer[1] == 1:
            pass
        else:
            answer[1] += move[0]
    elif a[i] == "R":
        if answer[1] == n:
            pass
        else:
            answer[1] += move[1]
    elif a[i] == "U":
        if answer[0] == 1:
            pass
        else:
            answer[0] += move[2]
    elif a[i] == "D":
        if answer[0] == n:
            pass
        else:
            answer[0] += move[3]
print(*answer)
```

```python
#시각 문제
n = int(input())

count = 0
h,m,s = 0,0,0
while h <= n:

    if s== 60:
        s=0
        m+=1
    if m == 60:
        m = 0
        h +=1
    if h > n:
        break

    if "3" in str(h) + str(s) + str(m):
        count +=1
    s+=1

print(count)
```

```python
#왕실의 나이트
row = [1,2,3,4,5,6,7,8]
col = ["a","b","c","d","e","f","g","h"]

night = [(-2,-1), (-1,-2),(-2,1),(-1,2),(2,-1),(1,-2),(2,1),(1,2)]


q = input().strip()

# 입력의 두 번째 문자를 정수로 바꿔 현재 행을 구합니다.
current_row = int(q[1])

# 열 문자의 인덱스에 1을 더해 a~h를 1~8로 바꿉니다.
current_col = col.index(q[0]) + 1

count = 0
for i in range(8):
    next_row = current_row + night[i][0]
    next_col = current_col + night[i][1]

    if 1 <= next_row <= 8 and 1 <= next_col <= 8:
            count += 1

# 가능한 이동의 개수를 출력합니다.
print(count)
```
