```python
# 테스트케이스 개수를 받습니다.
T = int(input())

# 테스트케이스를 하나씩 처리합니다.
for case in range(1, T + 1):
    # 두 숫자열의 길이를 받습니다.
    N, M = map(int, input().split())

    # 첫 번째 숫자열을 받습니다.
    A = list(map(int, input().split()))

    # 두 번째 숫자열을 받습니다.
    B = list(map(int, input().split()))

    # 항상 A가 짧은 숫자열이 되도록 맞춥니다.
    if N > M:
        # 숫자열 전체의 역할을 교환합니다. 원소 순서는 유지됩니다.
        A, B = B, A

        # 교환한 숫자열에 맞춰 길이도 바꿉니다.
        N, M = M, N

    # 각 배치에서 구한 곱의 합을 저장합니다.
    scores = []

    # 짧은 숫자열을 놓을 수 있는 시작 위치를 하나씩 확인합니다.
    for start in range(M - N + 1):
        # 새로운 배치의 합을 0부터 계산합니다.
        total = 0

        # 짧은 숫자열의 모든 원소를 확인합니다.
        for j in range(N):
            # 마주 보는 원소끼리 곱해서 합에 누적합니다.
            total += A[j] * B[start + j]

        # 현재 배치의 합을 저장합니다.
        scores.append(total)

    # 모든 배치의 합 중 최댓값을 선택합니다.
    answer = max(scores)

    # SWEA에서 요구하는 형식으로 출력합니다.
    print(f"#{case} {answer}")
```
