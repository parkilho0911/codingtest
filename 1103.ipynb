```python
#1989 복습 초심자의 회문 검사
t = int(input())

for start in range(1, t+1):
    n = input().strip()
    l = len(n)
    answer = 1
    for i in range(l//2):
        if n[i]!=n[l-1-i]:
            answer = 0
            break
    print(f"#{start} {answer}")
```

```python
#1970 복습 쉬운 거스름돈
t = int(input())
for start in range(1,t+1):
    n = int(input())
    money = [50000,10000,5000,1000,500,100,50,10]
    score = []
    for i in money:
        total = n//i
        score.append(total)
        n %= i

    print(f"#{start}\n", *score)
```

```python
#1926번 복습 간단한 369게임
n = int(input())
num = []
for i in range(1,n+1):
    count = 0
    for j in str(i):
        if j in "369":
            count +=1

    if count > 0:
        num.append("-"*count)
    else:
        num.append(i)
print(*num)
```

```python
#1284 수도 요금 경쟁
t = int(input())
#A사는 1리터당 P원을 낸다. 
#B사는 기본요금 Q이고 R리터를 넘지 않으면 Q로 청구된다. 하지만 R보다 많으면 1리터당 S원 받는다
for start in range(1,t+1):
    # P,Q,R,S,W순서이다
    #A사요금,기본요금,기준점,초과요금,사용한리터 
    # A사는 리터당 9원, 기본요금은 100 20리터 미만까지 넘으면 리터당 3원 추가 종민이는 W리터 사용
    P,Q,R,S,W = map(int, input().split())
    total = 0
    # 경우를 3개를 만들어야한다.
    #A사에 요금을 내는 경우, B사의 기본 요금을 내는 경우, B사의 기본요금 + 추가요금까지 내는 경우
    if W <= R: #A사 돈내기
        total = min(Q, P*W)
    elif W > R:
        A = P*W
        B = Q+(W-R)*S
        total = min(A, B)

    print(f"#{start} {total}")
```

```python
#1976 시각 덧셈
t = int(input())
for start in range(1,t+1):
    h1,m1,h2,m2 = map(int, input().split())
    min = m1+m2
    hous = h1+h2
    if min >= 60:
        hous +=1
        min -=60

    while hous > 12:
        hous -=12

    print(f"#{start} {hous} {min}")    
```

```python
#1948 날짜 계산기
t = int(input())
for start in range(1,t+1):
    m1,d1,m2,d2 = map(int, input().split())
    #지금 구해야하는 것은 처음 입력과 두번째 입력과의 날짜 차이를 게산하는 것이다.
    m_sum = 0
    if m1==m2:
        if m1==2:
            m_sum +=28-d1+1
        elif m1%2==0:
            m_sum += 30-d1+1
        else:
            m_sum +=31-d1+1

    for i in range(m2-m1):
        if m1+i==2:
            m_sum +=28-d1+1
        elif m1+i%2==0:
            m_sum += 30-d1+1
        else:
            m_sum +=31-d1+1

    print(f"#{start} {m_sum}")
```

```python
t = int(input())
for start in range(1,t+1):
    m1,d1,m2,d2 = map(int, input().split())
    days = [0,31,28,31,30,31,30,31,30,31,30,31,30]
    m_sum = 0
    if m1==m2: #여기 아예 달 계산을 통으로 해버려서 틀렸었음
        m_sum = d2-d1+1
    else:
        m_sum = days[m1] - d1 +1 #첫달 계산

        for month in range(m1+1, m2): #중간 달 계산
            m_sum += days[month]

        m_sum +=d2 #마지막 달 계산 
    print(f"#{start} {m_sum}")
```

```python
#1984번 중간 평균값 구하기
t = int(input())
for start in range(1, t+1):
    a = list(map(int, input().split()))
    a.sort()
    sum = 0
    count=0
    b = a[0]+a[-1]
    for i in a:
        sum +=i
        count+=1
    sum = sum-b
    count = count-2
    total = int(sum/count + 0.5) # 여기 반올림 부분을 몰라서 틀렸었음
    print(f"#{start} {total}")
```

```python
#1285번 아름이의 돌 던지기
#삼성에서는 C++ 밖에 없어서 제출 못함 지피티 결과 : 통과
#abs()가 양수로 바뀐다는 정보를 구글링함
t = int(input())

for start in range(1,t+1):
    n = int(input())
    a = list(map(int, input().split()))
    base = 0
    count = 0
    total = []
    for i in range(n):
        distance = base - a[i]
        total.append(distance)

    total = list(map(abs, total))
    total.sort()
    min_len = total[0]
    
    for j in total:
        if j ==min_len:
            count +=1
    print(f"#{start} {min_len} {count}")
```

```python
#1204번 1일차 - 최빈수 구하기
t = int(input())
for start in range(1,t+1):
    n = int(input())
    a = list(map(int, input().split()))

    count = [0]*101
    for i in a:
        count[i]+=1

    max_count = max(count)
    answer = 0
    for j in range(101):
        if count[j] == max_count:
            answer = j
    print(f"#{start} {answer}")
```

```python
#1288번 새로운 불면증 치료법
t = int(input())
for start in range(1,t+1):
    n = int(input())
    i = 0
    total = 0
    count = [0]*10
    while True:
        i +=n
        for j in str(i):
            count[int(j)]+=1
        if 0 not in count: # count not in 0: 이렇게해서 작동을 안했음. 값 not in 리스트 <<이게 정답            total = i
            total = i
            break

    print(f"#{start} {total}")
    
```
