### 1959번 다시 한번 복습하기

```python
T = int(input())

for start in range(1, T+1):
    N,M = map(int, input().split())

    A = list(map(int, input().split()))
    B = list(map(int, input().split()))

    if N > M: #지피티 도움 받음
        A,B=B,A
        N,M=M,N
    
    max_total=[]

    for i in range(M-N+1):
        total = 0
        for j in range(N):
            total += A[j]*B[i+j]
        max_total.append(total)
    
    maxsum=max(max_total)
    print(f"#{start} {maxsum}")
```

### 2001번 복습하기

```python
T = int(input())

for start in range(1, T+1):
    N,M = map(int, input().split())

    A = []
    for l in range(N):
        A.append(list(map(int, input().split())))

    sum_total=[]
    for i in range(N-M+1):
        for j in range(N-M+1):
            total = 0

            for row in range(M):
                for col in range(M):
                    total += A[i+row][j+col]
            sum_total.append(total)

    max_total = max(sum_total)

    print(f"#{start} {max_total}")
```

### 1945 간단한 소인수분해

```python
#insu배열하고 total>1이거를 지피티 도움 받음
T = int(input())

for start in range(1, T+1):
    N = int(input())

    total = N
    insu = [0]*5
    while total > 1:
        if total%11==0:
            total = total//11
            insu[4] += 1
        elif total%7==0:
            total = total//7
            insu[3] += 1
        elif total%5==0:
            total = total//5
            insu[2] += 1
        elif total%3==0:
            total = total//3
            insu[1] += 1
        else:
            total = total//2
            insu[0] += 1
    print(f"#{start}", *insu)
```

### 1986 지그재그 숫자

```python
T = int(input())

for start in range(1, T+1):
    N = int(input())
    total = 0
    i = 1
    for i in range(N+1):
        if i%2==0:
            total -=i
        else:
            total +=i

    print(f"#{start}", total)
```
