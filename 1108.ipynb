```python
#3499번 퍼펙트 셔플
t = int(input())

for test_case in range(1, t+1):
    n = int(input())
    a = list(map(str, input().split()))
    b = n // 2
    total = []
    x = []
    y = []
    # 1번 a리스트를 반으로 나눠서 각각 저장한다.
    # 2번 그 저장된 리스트를 b를 이용해서 꺼내온다.
    # 3번 그걸 그대로 total에 저장해서 출력한다.
    for i in range(2):
        if i == 0:
            if n % 2 == 0:
                for j in a[:b]:
                    x.append(j)
            else:
                for j in a[:b+1]:
                    x.append(j)
        else:
            if n % 2 == 0:
                for j in a[b:]:
                    y.append(j)
            else:
                for j in a[b+1:]:
                    y.append(j)
    if n%2==0:
        for k in range(b):
            total.append(x[k])
            total.append(y[k])
    else:
        for k in range(b+1):
            total.append(x[k])
            if k < len(y): # 마지막 y의 저장이 값이 없을때 범위를 어캐해야할지 몰라서 지피티에 물어봄
                total.append(y[k])
    print(f"#{test_case}",*total)

```

```python
#2805 농작물 수확하기
t = int(input())

for test_case in range(1,t+1):
    n = int(input())
    a = []
    total = 0
    half = n//2
    for i in range(n):
        a.append(list(map(int, input().strip())))
    # 1번 마름모를 구하는 방식을 찾아야함
    # 그래서 생각한게 n을 //2로 한 다음 for문을 2개 돌려서 하나는 위에서부터 반까지 가는거고
    # 하나는 밑에서부터 반 전까지 올라오는 것
    # 첫 행부터 가운데 행까지 확인합니다.
    for case1 in range(half + 1):

        # 현재 행에서 수확할 열의 시작과 끝을 구합니다.
        left = half - case1
        right = half + case1

        # right 열까지 포함하려고 끝값에 1을 더합니다.
        for col in range(left, right + 1):
            total += a[case1][col]  # 현재 행의 선택된 칸을 더합니다.

    for case2 in range(1, half + 1):

        left = 0 + case2
        right = n - 1 - case2

        for col in range(left, right + 1):
            total += a[half+case2][col]  # 현재 행의 선택된 칸을 더합니다.

    print(f"#{test_case} {total}")
```

```python
# 테스트케이스 10개를 처리합니다.
for test_case in range(1, 11):

    # 찾아야 하는 회문의 길이를 입력받습니다.
    n = int(input())

    # 8줄의 문자판을 저장할 리스트입니다.
    a = []

    # 이번 테스트케이스에서 발견한 회문의 개수입니다.
    answer = 0

    # 문자판의 각 줄을 문자 리스트로 만들어 저장합니다.
    for i in range(8):
        a.append(list(input().strip()))

    # 문자판의 첫 번째 행부터 마지막 행까지 확인합니다.
    for row in range(8):

        # 현재 행에서 n글자를 꺼낼 수 있는 시작 위치를 확인합니다.
        for col in range(8 - n + 1):

            # 현재 위치부터 n개의 문자를 꺼냅니다.
            word = a[row][col:col + n]

            # 순서를 뒤집어도 같으면 회문입니다.
            if word == word[::-1]:
                answer += 1

    # 모든 행의 가로 검사가 끝나면 결과를 출력합니다.
    print(f"#{test_case} {answer}")
```
