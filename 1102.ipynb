### #1945번 소인수분해 문제 복습

```python
#1945번 
T = int(input())

for start in range(1, T+1):
    N = int(input())

    insu = N
    total = [0]*5
    while N > 1:
        if N%11==0:
            N = N//11
            total[4] +=1
        elif N%7==0:
            N = N//7
            total[3] +=1
        elif N%5==0:
            N = N//5
            total[2] +=1
        elif N%3==0:
            N = N//3
            total[1] +=1
        else:
            N = N//2
            total[0] +=1

    print(f"#{start}",*total)
```

### 1959번 두 개의 숫자열

```python
# 1959 복습하기
T = int(input())

for start in range(1, T+1):
    N,M = map(int, input().split())

    A = list(map(int, input().split()))
    B = list(map(int, input().split()))

    score = []
    if N > M:
        N,M = M,N
        A,B = B,A

    for i in range(M-N+1):
        total = 0
        for j in range(N):
            total += A[j]*B[i+j]
        score.append(total)

    max_total=max(score)

    print(f"#{start}",max_total)
```

### 2001 파리 퇴치

```python
#2001번 파리 퇴치 복습
T = int(input())

for start in range(1, T+1):
    N,M = map(int, input().split())

    A = []
    score = []
    for _ in range(N):
        A.append(list(map(int, input().split())))

    for i in range(N-M+1):
        for j in range(N-M+1):
            total = 0
            for l in range(M):
                for z in range(M):
                    total += A[i+l][j+z]
            score.append(total)

    max_total = max(score)

    print(f"#{start}", max_total)
```

### 1989번 초심자의 회문 검사

```python
#1989번 회문 검사 복습
T = int(input())

for start in range(1,T+1):
    N = input().strip()

    A = len(N)
    answer = 1
    for i in range(A//2):
        if N[i]!=N[A-1-i]:
            answer = 0
            break

    print(f"#{start}",answer)

```

### 1970번 쉬운 거스름돈

```python
T = int(input())

for start in range(1,T+1):
    N = int(input())

    total = 0
    score =[]
    while N > 9:
        if N//50000>=1:
            total = N//50000
            score.append(total)
            N = N%50000
        elif N//10000>=1:
            total = N//10000
            score.append(total)
            N = N%10000
        elif N//5000>=1:
            total = N//5000
            score.append(total)
            N = N%5000
        elif N//1000>=1:
            total = N//1000
            score.append(total)
            N = N%1000
        elif N//500>=1:
            total = N//500
            score.append(total)
            N = N%500
        elif N//100>=1:
            total = N//100
            score.append(total)
            N = N%100
        elif N//50>=1:
            total = N//50
            score.append(total)
            N = N%50
        else:
            total = N//10
            score.append(total)
            N = N%10
    print(f"#{start}",*score)
```

```python
#지피티가 도와준 코드
T = int(input())

for start in range(1,T+1):
    n = int(input())
    
    money = [50000,10000,5000,1000,500,100,50,10]
    score = []

    for i in money:
        count = n//i
        score.append(count)
        n%=i
        
    print(f"#{start}\n",*score)
```

### 1926 간단한 369게임

```python
#1926 간단한 369게임
n = int(input())
score = []

for i in range(1,n+1):
    count = 0

    for j in str(i):
        if j in "369":
            count +=1

    if count == 0:
        score.append(str(i))
    else:
        score.append("-"*count)

print(*score)
```
