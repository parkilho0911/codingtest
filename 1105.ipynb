```python
#1940번 복습 가랏! RC카!
t = int(input())

for start in range(1,t+1):
    n = int(input())
    speed = 0
    distance = 0

    for i in range(n):
        a = list(map(int, input().split()))

        if a[0]==2:
            speed = max(0, speed-a[1])
        elif a[0]==0:
            pass
        elif a[0]==1:
            speed +=a[1]

        distance +=speed
    print(f"#{start} {distance}")
```

```python
#1946번 복습 간단한 압축 풀기
t = int(input())

for start in range(1,t+1):
    n = int(input())
    total = ""
    for i in range(n):
        a,b = input().split()
        b = int(b)
        total +=a*b

    print(f"#{start}")
    for i in range(0, len(total), 10):
        print(total[i:i+10])
```

```python
#2007 복습 패턴 마디의 길이
t = int(input())
for start in range(1,t+1):
    a = input().strip()
    answer = 0

    for j in range(1,11):
        pattern = a[:j]

        pattern2 = (pattern*len(a))[:len(a)]

        if pattern2 == a:
            answer = j
            break

    print(f"#{start} {answer}")
```

```python
#21425번 복습 +=
t = int(input())

for start in range(t):
    a,b,n = map(int, input().split())
    count = 0
    total = 0
    while total <= n:
        if a < b:
            a+=b
            total = a
        else:
            b+=a
            total = b
        count +=1
    print(count)
```

```python
#1204번 복습 최빈수 구하기
t = int(input())

for start in range(1,t+1):
    n = int(input())
    score = [0]*101 #100점까지의 칸
    student = list(map(int, input().split()))
    total = 0

    for i in student:
        score[i]+=1

    max_score = max(score)

    for j in range(101):
        if score[j] == max_score:
            total = j

    print(f"#{start} {total}")
```

```python
#1948번 복습 날짜 계산기
t = int(input())

for start in range(1,t+1):
    days = [0,31,28,31,30,31,30,31,31,30,31,30,31]
    m1,d1,m2,d2 = map(int, input().split())
    total = 0

    if m1==m2:
        total = d2-d1+1
    else:
        total = days[m1]-d1+1 #첫달 계산

        for i in range(m1+1, m2): #마지막 달 전까지 계산
            total +=days[i]

        total += d2

    print(f"#{start} {total}")
```

```python
#2005번 파스칼의 삼각형
t = int(input())
for start in range(1, t+1):
    n = int(input())
    total = []
    for i in range(n):
        row = [1]*(i+1)

        for j in range(1,i):
            row[j] = total[i-1][j-1] + total[i-1][j]

        total.append(row)
    print(f"#{start}")
    for k in range(n):
        print(*total[k])
```

```python
#1983번 조교의 성적 매기기
t = int(input())

for start in range(1,t+1):
    n,k = map(int,input().split())
    score = ["A+","A0","A-","B+","B0","B-","C+","C0","C-","D0"]
    total = [] #원본 데이터
    to_len = n//10

    for i in range(n):
        a,b,c = map(int,input().split())
        total.append(a * 0.35 + b * 0.45 + c * 0.2) #지피티 도움으로 고친거
    re = total[k-1]
    total.sort(reverse=True)
    rank = total.index(re)
    #여기까지가 등급하고 점수를 정렬을 마친상태
    for j in range(n):
        total[j] = score[j// to_len]
    student = total[rank]

    print(f"#{start} {student}")
```

```python
#1979번 어디에 단어가 들어갈 수 있을까
t = int(input())

for start in range(1,t+1):
    n,k = map(int, input().split())
    puzzle = []
    count = 0

    for i in range(n):
        puzzle.append(list(map(int, input().split())))

    #이제 내가 해야할게 가로 또는 세로가 k랑 같은 것을 찾아야한다.
    #가로버전 하나, 세로버전 하나
    for j in range(n):
        row = 0
        col = 0
        for z in range(n):
            if puzzle[j][z]==0:
                if row == k:
                    count +=1
                row = 0
            else:
                row += 1

            if puzzle[z][j]==0:
                if col ==k:
                    count +=1
                col = 0
            else:
                col += 1

        if row == k:
            count +=1

        if col == k:
            count +=1
    print(f"#{start} {count}")
```

```python
#1974번 스도쿠 검증
t = int(input())

for start in range(1,t+1):
    doku = []

    #스도쿠의 9개 행 입력
    for i in range(9):
        doku.append(list(map(int, input().split())))

    answer = 1 # 처음에는 올바른 스도쿠라고 가정

    #가로와 세로 검사
    for i in range(9):
        row_count = [0] * 10
        col_count = [0] * 10

        for j in range(9):
            row_count[doku[i][j]] +=1 #i행을 가로로 검사
            col_count[doku[j][i]] +=1 #i열을 세로로 검사

        if row_count[1:] !=[1]*9 or col_count[1:] !=[1]*9:
            answer = 0

    #3x3 구역의 시작 행과 열은 각각 0,3,6
    for block_row in range(0,9,3):
        for block_col in range(0,9,3):
            count = [0]*10 #구역이 바뀔 때마다 초기화

            #현재 구역 안의 3개 행과 3개 열 검사
            for row in range(block_row, block_row + 3):
                for col in range(block_col, block_col + 3):
                    count[doku[row[col]]] +=1

            if count[1:] != [1]*9:
                answer = 0
    print(f"#{start} {answer}")

```

```python
#1961번 숫자 배열 회전
t =int(input())
for start in range(1,t+1):
    n = int(input())
    row = []
    answer = []
    answer180 = []
    answer270 = []
    for i in range(n):
        row.append(list(map(int, input().split() )))

    for i in range(n): #90도 값
        line = []
        for j in range(n-1, -1, -1):#시작값, 끝값, 증가량
             line.append(row[j][i])

        answer.append(line)

    for i in range(n): #180도
        line = []

        for j in range(n-1,-1,-1):
            line.append(answer[j][i])
        answer180.append(line)

    for i in range(n): #270도
        line = []

        for j in range(n-1,-1,-1):
            line.append(answer180[j][i])
        answer270.append(line)

    print(f"#{start} ")
    for i in range(n):
        a90 = "".join(map(str, answer[i]))
        a180 = "".join(map(str, answer180[i]))
        a270 = "".join(map(str, answer270[i]))
        print(a90,a180,a270)






```

```python
# 1954번 달팽이 숫자
t = int(input())

for start in range(1, t + 1):
    n = int(input())

    # 숫자를 채울 n×n 배열 만들기
    answer = []
    for i in range(n):
        answer.append([0] * n)

    # 아직 채우지 않은 영역의 위·아래·왼쪽·오른쪽 경계
    top = 0
    bottom = n - 1
    left = 0
    right = n - 1

    num = 1  # 다음에 넣을 숫자

    # 채울 영역이 남아 있는 동안 반복
    while top <= bottom and left <= right:

        # ① 위쪽 줄: 왼쪽 → 오른쪽
        for col in range(left, right + 1):
            answer[top][col] = num
            num += 1
        top += 1  # 채운 위쪽 줄을 영역에서 제외

        # ② 오른쪽 줄: 위 → 아래
        for row in range(top, bottom + 1):
            answer[row][right] = num
            num += 1
        right -= 1  # 채운 오른쪽 줄을 영역에서 제외

        # ③ 아래쪽 줄: 오른쪽 → 왼쪽
        if top <= bottom:  # 채울 행이 남아 있을 때만 진행
            for col in range(right, left - 1, -1):
                answer[bottom][col] = num
                num += 1
            bottom -= 1  # 채운 아래쪽 줄을 영역에서 제외

        # ④ 왼쪽 줄: 아래 → 위
        if left <= right:  # 채울 열이 남아 있을 때만 진행
            for row in range(bottom, top - 1, -1):
                answer[row][left] = num
                num += 1
            left += 1  # 채운 왼쪽 줄을 영역에서 제외

    # 문제에서 요구하는 형식으로 한 행씩 출력
    print(f"#{start}")
    for line in answer:
        print(*line)
```
