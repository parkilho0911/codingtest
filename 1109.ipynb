```python
#1215번 복습 [S/W 문제해결 기본] 3일차 - 회문1
for test_case in range(1, 11):
    n = int(input())
    a = []
    count = 0
    for i in range(8):
        a.append(list(input().strip()))

    for i in range(8):
        for j in range(8-n+1): #가로를 구하는 것은 슬라이싱이 가능해 이렇게 작성
            word =  a[i][j:j+n]

            if word == word[::-1]:
                count+=1
        for k in range(8-n+1):
            word = []
            for row in range(k, k+n): # 세로는 슬라이싱이 아니라서 append활용
                word.append(a[row][i])

            if word == word[::-1]:
                count+=1

    print(f"#{test_case} {count}")
```

```python
# 13732번 정사각형 판정
t = int(input())

for test_case in range(1, t + 1):
    # n×n 문자판을 입력받습니다.
    n = int(input())
    a = []

    for i in range(n):
        a.append(list(input().strip()))

    # #이 있는 가장 위·아래·왼쪽·오른쪽 위치를 저장합니다.
    top = n
    bottom = -1
    left = n
    right = -1

    # 문자판 전체의 # 개수를 셉니다.
    total = 0

    for row in range(n):
        for col in range(n):
            if a[row][col] == "#":
                total += 1

                # #을 발견할 때마다 전체 범위를 갱신합니다.
                top = min(top, row)
                bottom = max(bottom, row)
                left = min(left, col)
                right = max(right, col)

    # 양 끝 칸을 모두 포함하므로 1을 더합니다.
    height = bottom - top + 1
    width = right - left + 1

    # 기본 판정은 정사각형이 아닌 것으로 둡니다.
    answer = "no"

    # #이 존재하고, 가로·세로 길이가 같으며,
    # 범위 안의 모든 칸이 #으로 채워져 있으면 정사각형입니다.
    if total > 0 and height == width and total == height * width:
        answer = "yes"

    # 테스트케이스마다 판정을 출력합니다.
    print(f"#{test_case} {answer}")
```
